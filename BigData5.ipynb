Exercice 1 — Logs système

```python
# EXERCICE 1 : Producer Kafka pour générer des logs système
# Les messages INFO sont envoyés vers logs_topic
# Les messages WARNING et ERROR sont envoyés vers alerts_topic

with open("producer_logs.py", "w") as f:
    f.write("""
from kafka import KafkaProducer
import random
import time

# Création du producer Kafka
producer = KafkaProducer(
    bootstrap_servers='localhost:9092',
    value_serializer=lambda v: v.encode('utf-8')
)

# Types de logs possibles
log_types = ["INFO", "WARNING", "ERROR"]

# Messages simulés
messages = [
    "Démarrage du service",
    "Connexion utilisateur réussie",
    "Utilisation mémoire élevée",
    "Erreur de connexion à la base de données",
    "Service arrêté",
    "Tentative d'accès non autorisée",
    "Fichier de configuration chargé"
]

# Génération de 20 logs
for i in range(20):
    log_type = random.choice(log_types)
    message = random.choice(messages)

    # Format demandé : INFO: ...
    log_message = f"{log_type}: {message}"

    # Routage selon le type du log
    if log_type == "INFO":
        producer.send("logs_topic", log_message)
        print(f"Envoyé vers logs_topic : {log_message}")
    else:
        producer.send("alerts_topic", log_message)
        print(f"Envoyé vers alerts_topic : {log_message}")

    # Petite pause pour simuler un flux temps réel
    time.sleep(1)

# Envoi définitif des messages
producer.flush()
producer.close()

print("Production des logs terminée.")
""")
```

Consumer 1 : lire logs_topic

```python
# Consumer 1 : il lit uniquement les logs INFO depuis logs_topic
# consumer_timeout_ms permet d'arrêter automatiquement le consumer après quelques secondes sans nouveau message

with open("consumer_logs.py", "w") as f:
    f.write("""
from kafka import KafkaConsumer

consumer = KafkaConsumer(
    "logs_topic",
    bootstrap_servers="localhost:9092",
    auto_offset_reset="earliest",
    enable_auto_commit=True,
    group_id="consumer_logs_group",
    consumer_timeout_ms=5000,
    value_deserializer=lambda v: v.decode("utf-8")
)

count_info = 0

print("Lecture des messages du topic logs_topic...")

for message in consumer:
    log = message.value
    print("Log reçu :", log)

    if log.startswith("INFO"):
        count_info += 1

consumer.close()

print("Nombre total de messages INFO :", count_info)
""")
```

```python
# Consumer 2 : il lit les alertes WARNING et ERROR depuis alerts_topic

with open("consumer_alerts.py", "w") as f:
    f.write("""
from kafka import KafkaConsumer

consumer = KafkaConsumer(
    "alerts_topic",
    bootstrap_servers="localhost:9092",
    auto_offset_reset="earliest",
    enable_auto_commit=True,
    group_id="consumer_alerts_group",
    consumer_timeout_ms=5000,
    value_deserializer=lambda v: v.decode("utf-8")
)

count_warning = 0
count_error = 0

print("Lecture des messages du topic alerts_topic...")

for message in consumer:
    alert = message.value
    print("Alerte reçue :", alert)

    if alert.startswith("WARNING"):
        count_warning += 1
    elif alert.startswith("ERROR"):
        count_error += 1

consumer.close()

print("Nombre total de WARNING :", count_warning)
print("Nombre total de ERROR :", count_error)
""")
```

Consumer général avec statistiques

```python
# Consumer général : il lit logs_topic et alerts_topic
# Il compte les messages INFO, WARNING et ERROR

with open("consumer_stats_logs.py", "w") as f:
    f.write("""
from kafka import KafkaConsumer
from collections import Counter

consumer = KafkaConsumer(
    "logs_topic",
    "alerts_topic",
    bootstrap_servers="localhost:9092",
    auto_offset_reset="earliest",
    enable_auto_commit=True,
    group_id="consumer_stats_group",
    consumer_timeout_ms=5000,
    value_deserializer=lambda v: v.decode("utf-8")
)

stats = Counter()

print("Lecture des deux topics logs_topic et alerts_topic...")

for message in consumer:
    log = message.value
    topic = message.topic

    print(f"[{topic}] {log}")

    log_type = log.split(":")[0]
    stats[log_type] += 1

consumer.close()

print("\\nStatistiques finales :")
print("INFO :", stats["INFO"])
print("WARNING :", stats["WARNING"])
print("ERROR :", stats["ERROR"])
""")
```

# ***Exercice 2 — Anomalies capteurs***

Producer capteurs

```python
# EXERCICE 2 : Producer Kafka pour générer des données de capteurs
# Format demandé : sensor1:20, sensor2:35, sensor3:100

with open("producer_sensors.py", "w") as f:
    f.write("""
from kafka import KafkaProducer
import random
import time

producer = KafkaProducer(
    bootstrap_servers="localhost:9092",
    value_serializer=lambda v: v.encode("utf-8")
)

sensors = ["sensor1", "sensor2", "sensor3"]

for i in range(30):
    sensor = random.choice(sensors)
    value = random.randint(10, 100)

    message = f"{sensor}:{value}"

    producer.send("sensor_topic", message)
    print("Donnée envoyée :", message)

    time.sleep(1)

producer.flush()
producer.close()

print("Production des données capteurs terminée.")
""")
```

Consumer anomalies + CSV

```python
# Consumer pour détecter les anomalies dans sensor_topic
# Une anomalie est une valeur supérieure au seuil 50
# Les données traitées sont enregistrées dans un fichier CSV

with open("consumer_sensors.py", "w") as f:
    f.write("""
from kafka import KafkaConsumer
from datetime import datetime
import csv

consumer = KafkaConsumer(
    "sensor_topic",
    bootstrap_servers="localhost:9092",
    auto_offset_reset="earliest",
    enable_auto_commit=True,
    group_id="consumer_sensors_group",
    consumer_timeout_ms=5000,
    value_deserializer=lambda v: v.decode("utf-8")
)

threshold = 50

total_values = 0
sum_values = 0
total_anomalies = 0

csv_filename = "sensor_data.csv"

with open(csv_filename, mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    # En-tête du fichier CSV
    writer.writerow(["date", "sensor", "value", "status"])

    print("Lecture des données capteurs...")

    for message in consumer:
        data = message.value

        # Extraction du capteur et de la valeur
        sensor, value = data.split(":")
        value = int(value)

        # Mise à jour des statistiques
        total_values += 1
        sum_values += value
        average = sum_values / total_values

        # Détection d'anomalie
        if value > threshold:
            status = "ANOMALIE"
            total_anomalies += 1
            print(f"ALERTE : anomalie détectée sur {sensor}, valeur = {value}")
        else:
            status = "NORMAL"

        # Écriture dans le fichier CSV
        writer.writerow([
            datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            sensor,
            value,
            status
        ])

        print(f"Donnée traitée : {sensor} = {value}, statut = {status}")
        print("Moyenne actuelle :", round(average, 2))
        print("Nombre total d'anomalies :", total_anomalies)
        print("--------------------------------")

consumer.close()

print("Fichier CSV créé :", csv_filename)
""")
```

# ***Exercice 3 — Pipeline profils + transactions***

Producer profils utilisateurs

```python
# EXERCICE 3 : Producer pour envoyer les profils utilisateurs vers users_topic
# Format demandé : user1:premium, user2:standard, user3:premium

with open("producer_users.py", "w") as f:
    f.write("""
from kafka import KafkaProducer
import time

producer = KafkaProducer(
    bootstrap_servers="localhost:9092",
    value_serializer=lambda v: v.encode("utf-8")
)

users = [
    "user1:premium",
    "user2:standard",
    "user3:premium"
]

for user in users:
    producer.send("users_topic", user)
    print("Profil envoyé :", user)
    time.sleep(1)

producer.flush()
producer.close()

print("Production des profils terminée.")
""")
```

Producer transactions

```python
# Producer pour envoyer les transactions vers transactions_topic
# Format demandé : user1:200, user2:50, user3:1000

with open("producer_transactions.py", "w") as f:
    f.write("""
from kafka import KafkaProducer
import time

producer = KafkaProducer(
    bootstrap_servers="localhost:9092",
    value_serializer=lambda v: v.encode("utf-8")
)

transactions = [
    "user1:200",
    "user2:50",
    "user3:1000",
    "user2:700",
    "user1:2500"
]

for transaction in transactions:
    producer.send("transactions_topic", transaction)
    print("Transaction envoyée :", transaction)
    time.sleep(1)

producer.flush()
producer.close()

print("Production des transactions terminée.")
""")
```

Consumer pipeline temps réel

```python
# Consumer pipeline :
# 1. lit users_topic et transactions_topic
# 2. stocke les profils utilisateurs en mémoire
# 3. enrichit les transactions
# 4. détecte les transactions suspectes selon le type d'utilisateur

with open("consumer_pipeline.py", "w") as f:
    f.write("""
from kafka import KafkaConsumer

consumer = KafkaConsumer(
    "users_topic",
    "transactions_topic",
    bootstrap_servers="localhost:9092",
    auto_offset_reset="earliest",
    enable_auto_commit=True,
    group_id="consumer_pipeline_group",
    consumer_timeout_ms=5000,
    value_deserializer=lambda v: v.decode("utf-8")
)

# Dictionnaire pour stocker les profils utilisateurs
# Exemple : {"user1": "premium", "user2": "standard"}
user_profiles = {}

# Seuils selon le type d'utilisateur
thresholds = {
    "standard": 500,
    "premium": 2000
}

print("Lecture des profils et transactions...")

for message in consumer:
    topic = message.topic
    data = message.value

    # Cas 1 : message venant du topic des profils
    if topic == "users_topic":
        user_id, user_type = data.split(":")

        user_profiles[user_id] = user_type

        print(f"Profil enregistré : {user_id} est {user_type}")

    # Cas 2 : message venant du topic des transactions
    elif topic == "transactions_topic":
        user_id, amount = data.split(":")
        amount = int(amount)

        # Récupération du type utilisateur
        user_type = user_profiles.get(user_id, "unknown")

        # Transaction enrichie
        print(f"Transaction enrichie : {user_id} ({user_type}) : {amount}")

        if user_type == "unknown":
            print("Profil inconnu : impossible d'appliquer un seuil précis.")

        else:
            threshold = thresholds[user_type]

            if amount > threshold:
                print(f"ALERTE : transaction suspecte pour {user_id}")
                print(f"Montant = {amount}, seuil autorisé = {threshold}")
            else:
                print("Transaction normale.")

    print("--------------------------------")

consumer.close()
""")
```

# Conclusion

Dans ce TP, nous avons utilisé Apache Kafka pour traiter des flux de données en temps réel.

Dans le premier exercice, nous avons créé un producer qui génère des logs système de type INFO, WARNING et ERROR. Les logs INFO sont envoyés vers le topic `logs_topic`, tandis que les logs WARNING et ERROR sont envoyés vers le topic `alerts_topic`. Des consumers permettent ensuite de lire ces messages et de calculer des statistiques.

Dans le deuxième exercice, nous avons simulé des données de capteurs envoyées vers Kafka. Le consumer lit les valeurs, détecte les anomalies lorsque la valeur dépasse le seuil 50, calcule la moyenne des valeurs reçues et enregistre les données dans un fichier CSV.

Dans le troisième exercice, nous avons conçu un pipeline temps réel avec deux flux : les profils utilisateurs et les transactions. Le consumer stocke les profils en mémoire, enrichit chaque transaction avec le type d'utilisateur correspondant et détecte les transactions suspectes selon un seuil adapté au profil utilisateur.

Ce TP montre que Kafka est utile pour la supervision, la détection d'anomalies et le traitement de données en temps réel.
